# Better Understanding the Data

The *music_info* dataframe contains 50683 rows, each corresponding to a unique song found on Spotify, with 21 columns of identifying and descriptory information. The first column, *track_id*, serves to uniquely label each entry.

The *listening_info* dataframe contains >9 million rows, each corresponding to a unique combination of a *track_id* column and *user_id* column, followed by the value *playcount*. This allows us to relate any indivual *user_id* with the unique *track_id* list they have listened to, or to relate a *track_id* to both unique and total listeners.

```python
music_info.info()
```

```python
listening_info.info()
```

By joining these two DataFrames, using *track_id* as our index, we can make relational analyses.

```python
left = music_info.set_index('track_id')
right = listening_info.set_index('track_id')
music_listening_info = left.join(right)
music_listening_info.head()
```

# Interesting Ways to Search:

1. Searching by Artist

```python
def get_discography(artist_name):
    discography = music_info.loc[music_info.artist == artist_name]
    return discography

search = get_discography('Radiohead')
search.head()
```

2. Searching by tags

```python
def get_tags(tags):
    search = music_info.loc[music_info.tags.isin(tags)]
    #search = music_info.tags.str.contains(tags)
    return search

search = get_tags(['electronic', 'alternative'])
search.head()
```

3. Searching by track_id

```python
def get_song_info(id):
    artist = music_info.loc[music_info.track_id == id].artist
    name = music_info.loc[music_info.track_id == id].name
    info = pd.merge(artist, name, right_index=True, left_index=True)
    return info

track_id = 'TRLNZBD128F935E4D8'
get_song_info(track_id)
```

# Looking at Trends:

As one of the most basic and straightforward relationships between the two databases, let's first look at the most popular tracks in the database according to user listening history.

**Top Ten Tracks by Listen History**

The *count* lists the number of *user_id* entries associated with each *track_id* (i.e. number of unique listeners). The *sum* lists the summed total playcount of each track.

```python
track_id_playcount = listening_info.groupby('track_id').playcount.agg(['count', 'sum'])
top_ten_id = track_id_playcount.nlargest(10, 'sum')
#top_ten_tracks = top_ten_id.apply(lambda n : get_song_info(n))
print(top_ten_id)
```
